## Cleaning Data 

```python
clean_data = data.copy()
```

```python
clean_data.isnull().sum()
```

```python
# Handle Missing Customer_ID
```

```python
clean_data[clean_data["Customer_ID"].isnull()]
```

```python
clean_data["Customer_ID"] = clean_data["Customer_ID"].fillna("Unknown")
```

```python
clean_data["Customer_ID"].isnull().sum()
```

```python
# Handle Missing Payment Method
```

```python
clean_data["Payment_Method"].value_counts(dropna=False)
```

```python
clean_data["Payment_Method"] = clean_data["Payment_Method"].fillna(
    clean_data["Payment_Method"].mode()[0]
)
```

```python
clean_data["Payment_Method"].isnull().sum()
```

```python
# Handle Missing Price
```

```python
clean_data[clean_data["Price"].isnull()]
```

```python
clean_data["Price"] = clean_data["Price"].fillna(
    clean_data.groupby("Item")["Price"].transform("median")
)
```

```python
clean_data["Price"] = clean_data["Price"].fillna(
    clean_data["Price"].median()
)
```

```python
clean_data["Price"].isnull().sum()
```

```python
clean_data.isnull().sum()
```

```python
clean_data[clean_data["Order_Date"].isnull()]
```

```python
data.loc[clean_data["Order_Date"].isna(), "Order_Date"].head(20)
```

```python
clean_data["Order_Date"] = pd.to_datetime(
    clean_data["Order_Date"],
    errors="coerce",
    dayfirst=False
)
```

```python
clean_data["Order_Date"].isnull().sum()
```

```python
data.dtypes
```

```python
median_date = clean_data["Order_Date"].dropna().median()

clean_data["Order_Date"] = clean_data["Order_Date"].fillna(median_date)
```

```python
clean_data["Order_Date"].isnull().sum()
```

```python
clean_data.duplicated().sum()
```

```python
clean_data = clean_data.drop_duplicates()
```

```python
clean_data.duplicated().sum()
```

```python
clean_data.shape
```

```python
data.head()
```

```python
clean_data.select_dtypes(include="object").columns
```

```python
text_cols = clean_data.select_dtypes(include="object").columns

for col in text_cols:
    clean_data[col] = clean_data[col].str.strip()
```

```python
data.head()
```

```python
clean_data["Payment_Method"].unique()
```

```python
clean_data["Payment_Method"].value_counts()
```

```python
clean_data["Payment_Method"] = clean_data["Payment_Method"].replace({
    "CARD": "Credit Card",
    "card": "Credit Card",
    "cash": "Cash",
    "upi": "Digital Wallet"
})
```

```python
clean_data["Payment_Method"].value_counts()
```

```python
clean_data["Payment_Method"] = clean_data["Payment_Method"].replace(
    "Unknown",
    np.nan
)
```

## Checking Outliers

```python
clean_data[clean_data["Price"] > 1500]
```

```python
clean_data.loc[
    clean_data["Price"] <= 0,
    "Price"
] = np.nan
```

```python
clean_data["Price"] = clean_data["Price"].fillna(
    clean_data.groupby("Item")["Price"].transform("median")
)
```

```python
clean_data[clean_data["Quantity"] <= 0]
```

```python
clean_data[clean_data["Quantity"] > 20]
```

```python
clean_data.loc[
    clean_data["Quantity"] <= 0,
    "Quantity"
] = 1
```

```python
clean_data.loc[
    clean_data["Quantity"] > 20,
    "Quantity"
] = 5
```

```python
clean_data["Order_Total"] = (
    clean_data["Price"] *
    clean_data["Quantity"]
)
```
